# Part 2

Here, things get even more interesting! We will compare the major interest at enrollement and within the 4 years of college for retained students.

This step involves using mapping pipelines to obtain program codes ie(BIOL for Biology) for listed enrollment interest because, as stated earlier, enrollment interest is in the form of simple program titles.

We need them to be in the same format because the colum of graduation majors and swithced majors is in from of program codes.


To achieve this we'll use a library called fuzzy wuzzy. A limtation of this library for our task is that it does character matching which is inefficient hre since our analysis requires the program to identify the meanings and not just character match the strings. To create a structured and constrained pipeline to ensure maximum accuracy of our program we follow these steps:

1. Normalization
2. Creating a controlled vocabulary map
3. Passing the lists of strings through fuzzy pipeline and mapping dictionary



1. Normalization

```python
#importing and evaluating the format of program codes dataset
program_codes = pd.read_csv('/content/drive/MyDrive/CS215-Spring-ESTHER/Projects/Final Project/Enrolled Applicants fall 2017-2021.xlsx - Program Codes and Titles.csv')
type(program_codes['Program Title'].iloc[0])
program_codes['Program Title'].iloc[0]
type(enrollement_and_grad['All Academic Interests'].iloc[0][0])
type(program_codes['Program Title'].iloc[0])
#int_dec.columns
```

```python
#we need some normalization
import re #this module cleans data by removing unwanted text
def normalize(text):
  text = text.lower()
  text = re.sub(r"[^a-z0-9\s]", "", text) # remove punctuations
  text = re.sub(r"\s+", " ", text).strip() #fix spacing
  return text

program_codes['normalized_titles'] = program_codes['Program Title'].apply(normalize)
enrollement_and_grad['normalized_interests'] = enrollement_and_grad['All Academic Interests'].apply(lambda x: [normalize(interest) for interest in x])
```

2. Creating a controlled vocabulary map

```python
# Create mapping dictionary
interests = enrollement_and_grad['normalized_interests']
program_map = dict(
    zip(
        program_codes['normalized_titles'],
        program_codes['Program Code']
    )
)
```

3. Mapping lists of strings to

```python
# Install fuzzywuzzy library for fuzzy string matching
#Allows to match the subjects despite the inconsistent naming order compared to the program code titles

!pip install fuzzywuzzy
!pip install python-Levenshtein
```

```python
from fuzzywuzzy import process

def  match_interest_to_code(interests_list, program_map, threshold=50):
    matched_codes = []
    for interest in interests_list:
        # Find the best matching program title for the given interest
        # using fuzzy matching. process.extractOne returns (match, score)
        match = process.extractOne(interest, program_map.keys(), score_cutoff=threshold)
        if match:
            # If a match is found above the threshold, add its corresponding code
            matched_codes.append(program_map[match[0]])
    return list(set(matched_codes)) # Return unique codes

# Apply the fuzzy matching function to create new 'fuzzy interest codes'
enrollement_and_grad['fuzzy_interest_codes'] = enrollement_and_grad['normalized_interests'].apply(
    lambda x: match_interest_to_code(x, program_map)
)

# Display the results with both normalized interests and the new fuzzy codes
enrollement_and_grad[['normalized_interests',  'fuzzy_interest_codes', 'Program Code']]
type(enrollement_and_grad['fuzzy_interest_codes'].iloc[0])
```

```python
#removing the first two characters BA in Program codes
enrollement_and_grad['Program Code'] = enrollement_and_grad['Program Code'].str.strip('BA.')
enrollement_and_grad[['normalized_interests',  'fuzzy_interest_codes', 'Program Code', 'count']].head(5)
```

2. Examining what programs the students swithced to and from

```python
changed_mind['Majors at Graduation'].count()
```

```python
import numpy as np

# Normalize 'Program Code' and 'Majors at Graduation' in changed_mind for consistent comparison
changed_mind_normalized = changed_mind.copy()
changed_mind_normalized['Program Code'] = changed_mind_normalized['Program Code'].str.strip('BA.')
changed_mind_normalized['Majors at Graduation'] = changed_mind_normalized['Majors at Graduation'].str.strip('BA.')

def count_students_in_grad_programs(current_program_code):
  # This 'current_program_code' is a single value from program_codes['Program Code']
  # We want to count how many times this specific program_code appears in first_decisions['Program Code']
  return (changed_mind_normalized['Majors at Graduation'] == current_program_code).sum()


def count_students_in_programs_at_first_choice(current_program):
  return (changed_mind_normalized['Program Code'] == current_program).sum()

program_codes['first_choice_programs'] = program_codes['Program Code'].apply(count_students_in_programs_at_first_choice)
program_codes['grad_majors_count'] = program_codes['Program Code'].apply(count_students_in_grad_programs)
program_codes['diff'] = program_codes['first_choice_programs'] - program_codes['grad_majors_count']

# Calculate percentage_change, handling division by zero using np.where
# If 'first_choice_programs' is 0, the percentage change is 0 (no initial students to change from)
program_codes['percentage_change'] = np.where(
    program_codes['first_choice_programs'] != 0,
    (program_codes['diff'] / program_codes['first_choice_programs']) * 100, 0
)

# Further ensure NaN or inf values resulting from edge cases are set to 0
program_codes['percentage_change'] = program_codes['percentage_change'].replace([np.inf, -np.inf], 0).fillna(0)

program_codes_second_sort = program_codes.sort_values(by = 'first_choice_programs', ascending = False)
program_codes_second_sort
```

```python
null_codes = np.where((program_codes_second_sort['grad_majors_count'] == 0) & (program_codes_second_sort['first_choice_programs'] == 0) )[0] # retunrs the array insidethe tuple generated by np.where()
print(f"Number of programs with no change in student count {len(null_codes)}")

b_and_a_student_count = program_codes_second_sort.drop(null_codes)
```

Determining the rate of graduation for students who never switched majors

```python
# Filter for students who have graduated
graduated_students = enrollement_and_grad[enrollement_and_grad['Student Status'] == 'Graduated'].copy()

# Normalize 'Majors at Graduation' by stripping 'BA.' for consistent comparison
graduated_students['normalized_grad_major'] = graduated_students['Majors at Graduation'].str.strip('BA.').fillna('')

# Define a function to check if the graduated major is in the initial fuzzy interests
def stuck_with_interest(row):
    grad_major = row['normalized_grad_major']
    fuzzy_interests = row['fuzzy_interest_codes']
    return grad_major in fuzzy_interests and grad_major != ''

# Apply the function to count students who stuck with their interest
students_stuck_with_first_year_interest = graduated_students['normalized_grad_major'].apply(lambda x: x if x in program_map.keys() else None).dropna().index.size

# Count students where the final graduated major was among their initial fuzzy interests
count_stuck = graduated_students.apply(stuck_with_interest, axis=1).sum()

# Calculate the total number of graduated students for percentage
total_graduated_students = len(graduated_students)

# Calculate the percentage
percentage_stuck = (count_stuck / total_graduated_students) * 100 if total_graduated_students > 0 else 0

print(f"Out of {total_graduated_students} graduated students, {count_stuck} stuck with one of their first-year academic interests as their graduation major.")
print(f"This is {percentage_stuck:.2f}% of all graduated students.")
```
